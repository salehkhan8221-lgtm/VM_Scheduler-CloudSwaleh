```python
# ========== SECTION 1: IMPORT REQUIRED LIBRARIES ==========
print("📦 Setting up environment...")

# System and data libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import time
import logging
import sys

# Setup logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

# Import optimized modules
sys.path.insert(0, str(Path('.').absolute()))

from data_processor import DataProcessor
from model_manager import ModelManager
from vm_allocator import VMAllocator, HostResource
from metrics_tracker import MetricsTracker

print("✅ Environment setup complete!\n")
```

## Section 2: Performance Optimization - Data Loading & Processing
Load data efficiently with memory optimization and preprocessing

```python
# ========== OPTIMIZED DATA LOADING WITH MEMORY EFFICIENCY ==========
print("=" * 70)
print("OPTIMIZED DATA PROCESSING")
print("=" * 70)

# Initialize data processor
processor = DataProcessor()

# Simulate creating a sample dataset if file doesn't exist
dataset_path = "Dataset/Dataset with timestamp.csv"

if Path(dataset_path).exists():
    print(f"\n📥 Loading dataset from {dataset_path}...")
    
    # Measure loading time
    start_time = time.time()
    df = processor.load_dataset(dataset_path)
    load_time = time.time() - start_time
    
    print(f"⏱️  Loading time: {load_time:.3f} seconds")
    print(f"📊 Dataset shape: {df.shape}")
    print(f"📝 Columns: {list(df.columns)}")
    print(f"💾 Memory usage: {df.memory_usage(deep=True).sum() / 1024 / 1024:.2f} MB")
else:
    print(f"\n⚠️  Dataset not found at {dataset_path}")
    print("Creating synthetic dataset for demonstration...")
    
    # Create synthetic data
    timestamps = pd.date_range('2023-01-01', periods=10000, freq='15min')
    cpu_usage = 40 + 30 * np.sin(np.arange(10000) / 100) + np.random.normal(0, 5, 10000)
    cpu_usage = np.clip(cpu_usage, 0, 100)
    
    df = pd.DataFrame({
        'Timestamp': timestamps,
        'CPUUsage': cpu_usage
    })
    
    processor.df = df
    print(f"✅ Synthetic dataset created: {df.shape}")

# Process data
print("\n🔧 Processing data...")
processor.standardize_columns()
processor.format_timestamps()
processor.extract_time_features()

# Calculate statistics
stats = processor.calculate_statistics()
print(f"\n📈 CPU Usage Statistics:")
print(f"   Mean: {stats['mean']:.2f}%")
print(f"   Std Dev: {stats['std']:.2f}%")
print(f"   Min: {stats['min']:.2f}%")
print(f"   Max: {stats['max']:.2f}%")
```

## Section 3: ML Model Training with Caching

Train and evaluate Linear Regression model with performance metrics

```python
# ========== ML MODEL TRAINING WITH CACHING ==========
print("\n" + "=" * 70)
print("ML MODEL TRAINING & OPTIMIZATION")
print("=" * 70)

# Initialize model manager
model_manager = ModelManager()

# Create sequences from data
sequence_length = 10
print(f"\n📊 Preparing data (sequence_length={sequence_length})...")

start_prep = time.time()
X, y = processor.create_sequences(sequence_length)
prep_time = time.time() - start_prep

print(f"⏱️  Preparation time: {prep_time:.3f} seconds")
print(f"📈 Sequences created: X shape {X.shape}, y shape {y.shape}")

# Prepare data for training
print(f"\n🔧 Scaling and splitting data...")

start_train_prep = time.time()
X_train, X_test, y_train, y_test = model_manager.prepare_data(X, y, test_size=0.2)
train_prep_time = time.time() - start_train_prep

print(f"⏱️  Training preparation time: {train_prep_time:.3f} seconds")
print(f"📊 Training set: {X_train.shape}")
print(f"📊 Test set: {X_test.shape}")

# Train Linear Regression model
print(f"\n🤖 Training Linear Regression model...")

start_train = time.time()
model_manager.train_linear_regression(X_train, y_train)
train_time = time.time() - start_train

print(f"⏱️  Training time: {train_time:.3f} seconds")
print(f"✅ Model trained successfully!")

# Evaluate model
print(f"\n📈 Evaluating model performance...")

start_eval = time.time()
metrics = model_manager.evaluate(X_test, y_test)
eval_time = time.time() - start_eval

print(f"⏱️  Evaluation time: {eval_time:.3f} seconds")
print(f"\n🎯 Model Metrics:")
print(f"   R² Score: {metrics['r2']:.4f}")
print(f"   RMSE: {metrics['rmse']:.4f}")
print(f"   MAE: {metrics['mae']:.4f}")
print(f"   MSE: {metrics['mse']:.4f}")

# Save model for later use
print(f"\n💾 Saving model to cache...")
model_manager.save_model()
print(f"✅ Model cached successfully")
```

## Section 4: CPU/Storage Prediction & VM Allocation with Resource Constraints

```python
# ========== CPU PREDICTION & VM ALLOCATION ==========
print("\n" + "=" * 70)
print("CPU PREDICTION & VM ALLOCATION")
print("=" * 70)

# Generate predictions for next N values
pred_steps = 10
print(f"\n🔮 Generating {pred_steps} future CPU predictions...")

start_pred = time.time()
last_sequence = X_train[-1]  # Get last sequence from training data
future_predictions = model_manager.predict_next_values(
    last_sequence, 
    num_steps=pred_steps
)
pred_time = time.time() - start_pred

print(f"⏱️  Prediction time: {pred_time:.3f} seconds")
print(f"\n📊 Future CPU Predictions:")
for i, pred in enumerate(future_predictions, 1):
    print(f"   Step {i}: {pred:.2f}%")

# Initialize VM Allocator with resource constraints
print(f"\n🖥️  Setting up Datacenter with Resource Constraints...")

allocator = VMAllocator(
    num_hosts=10,
    cpu_per_host=8.0,      # 8 cores per host
    memory_per_host=16384,  # 16 GB per host
    storage_per_host=500    # 500 GB per host
)

allocator.set_thresholds(cpu=80, memory=85, storage=80)

print(f"✅ Datacenter configured:")
print(f"   Hosts: {len(allocator.hosts)}")
print(f"   CPU per host: {allocator.hosts[0].total_cpu_cores} cores")
print(f"   Memory per host: {allocator.hosts[0].total_memory} MB")
print(f"   Storage per host: {allocator.hosts[0].total_storage} GB")

# Perform VM allocation based on predictions
print(f"\n⚡ Allocating VMs based on CPU predictions...")

start_alloc = time.time()
allocation_result = allocator.allocate_predictive(future_predictions)
alloc_time = time.time() - start_alloc

print(f"⏱️  Allocation time: {alloc_time:.3f} seconds")
print(f"\n📊 Allocation Results:")
print(f"   Total VMs processed: {allocation_result['total_vms']}")
print(f"   Successfully allocated: {allocation_result['successful']}")
print(f"   Failed allocations: {allocation_result['failed']}")
success_rate = (allocation_result['successful'] / allocation_result['total_vms'] * 100)
print(f"   Success rate: {success_rate:.1f}%")

# Get datacenter utilization
print(f"\n📈 Datacenter Utilization:")
util = allocator.get_datacenter_utilization()
print(f"   CPU Utilization: {util['cpu_utilization']:.2f}%")
print(f"   Memory Utilization: {util['memory_utilization']:.2f}%")
print(f"   Storage Utilization: {util['storage_utilization']:.2f}%")
print(f"   Total VMs: {util['total_vms']}")
print(f"   Hosts with available resources: {util['hosts_with_available_resources']}/{len(allocator.hosts)}")
```

## Section 5: Real-Time Metrics Tracking & Monitoring

Track system performance metrics, alerts, and health status in real-time

```python
# ========== REAL-TIME METRICS TRACKING ==========
print("\n" + "=" * 70)
print("REAL-TIME METRICS TRACKING & MONITORING")
print("=" * 70)

# Initialize metrics tracker
tracker = MetricsTracker(max_history=1000)

# Simulate recording metrics over time
print(f"\n📊 Recording metrics over {len(future_predictions)} time steps...")

start_tracking = time.time()

for step, pred_cpu in enumerate(future_predictions):
    # Create realistic metrics based on predictions
    cpu_usage = pred_cpu + np.random.normal(0, 2)
    memory_usage = 50 + (pred_cpu / 100) * 30 + np.random.normal(0, 5)
    storage_usage = 60 + np.random.normal(0, 3)
    vms_count = int(pred_cpu / 10)
    active_hosts = max(1, int(vms_count / 3))
    
    # Record metric
    tracker.record_metric(
        timestamp=step,
        cpu_usage=np.clip(cpu_usage, 0, 100),
        memory_usage=np.clip(memory_usage, 0, 100),
        storage_usage=np.clip(storage_usage, 0, 100),
        vms_count=vms_count,
        active_hosts=active_hosts,
        prediction=pred_cpu,
        allocation_success=allocation_result['successful'] > 0
    )

tracking_time = time.time() - start_tracking
print(f"⏱️  Tracking time: {tracking_time:.3f} seconds")

# Get summary statistics
print(f"\n📈 Summary Statistics:")
summary = tracker.get_summary()

print(f"\nCPU Usage:")
print(f"   Mean: {summary['cpu']['mean']:.2f}%")
print(f"   Max: {summary['cpu']['max']:.2f}%")
print(f"   Min: {summary['cpu']['min']:.2f}%")
print(f"   Std Dev: {summary['cpu']['std']:.2f}%")

print(f"\nMemory Usage:")
print(f"   Mean: {summary['memory']['mean']:.2f}%")
print(f"   Max: {summary['memory']['max']:.2f}%")

print(f"\nStorage Usage:")
print(f"   Mean: {summary['storage']['mean']:.2f}%")
print(f"   Max: {summary['storage']['max']:.2f}%")

print(f"\nVMs Statistics:")
print(f"   Average VMs: {summary['vms']['mean']:.1f}")
print(f"   Max VMs: {summary['vms']['max']}")

print(f"\n🎯 System Health Metrics:")
print(f"   Performance Index: {tracker.get_performance_index():.1f}/100")
print(f"   Health Status: {tracker.get_health_status()}")
print(f"   Total Records: {summary['total_records']}")

# Display alerts
alerts = tracker.get_alerts()
print(f"\n⚠️  System Alerts: {len(alerts)}")
if alerts:
    for alert in alerts[:3]:  # Show first 3 alerts
        print(f"   - {alert['message']}")
```

## Section 6: Performance Visualization & Analysis

```python
# ========== PERFORMANCE VISUALIZATION ==========
print("\n" + "=" * 70)
print("GENERATING PERFORMANCE VISUALIZATIONS")
print("=" * 70)

# Create figure with subplots
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('VM Scheduler CloudSim - Performance Analysis Dashboard', fontsize=16, fontweight='bold')

# 1. CPU Predictions
recent_metrics = tracker.get_recent_metrics(10)
ax1 = axes[0, 0]
ax1.plot(recent_metrics['timestamp'], recent_metrics['cpu_usage'], 'b-o', label='Actual CPU', linewidth=2)
ax1.plot(range(len(future_predictions)), future_predictions, 'r--s', label='Predicted CPU', linewidth=2)
ax1.set_xlabel('Time Step')
ax1.set_ylabel('CPU Usage (%)')
ax1.set_title('CPU Usage: Actual vs Predicted')
ax1.legend()
ax1.grid(True, alpha=0.3)

# 2. Resource Utilization
ax2 = axes[0, 1]
resources = ['CPU', 'Memory', 'Storage']
utilization = [
    util['cpu_utilization'],
    util['memory_utilization'],
    util['storage_utilization']
]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
bars = ax2.bar(resources, utilization, color=colors, alpha=0.7, edgecolor='black')
ax2.set_ylabel('Utilization (%)')
ax2.set_title('Datacenter Resource Utilization')
ax2.set_ylim(0, 100)
for bar, util_val in zip(bars, utilization):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
            f'{util_val:.1f}%', ha='center', va='bottom', fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y')

# 3. VM Allocation Success
ax3 = axes[0, 2]
categories = ['Successful', 'Failed']
values = [allocation_result['successful'], allocation_result['failed']]
colors_alloc = ['#2ca02c', '#d62728']
ax3.pie(values, labels=categories, autopct='%1.1f%%', colors=colors_alloc, startangle=90)
ax3.set_title(f'VM Allocation Results\n(Total: {allocation_result["total_vms"]} VMs)')

# 4. Host Statistics
ax4 = axes[1, 0]
host_stats = allocator.get_host_statistics()
ax4.bar(host_stats['host_id'], host_stats['vms_count'], color='#1f77b4', alpha=0.7, edgecolor='black')
ax4.set_xlabel('Host ID')
ax4.set_ylabel('Number of VMs')
ax4.set_title('VMs per Host')
ax4.grid(True, alpha=0.3, axis='y')

# 5. System Metrics Over Time
ax5 = axes[1, 1]
ax5.plot(recent_metrics['timestamp'], recent_metrics['cpu_usage'], 'b-o', label='CPU', linewidth=2)
ax5.plot(recent_metrics['timestamp'], recent_metrics['memory_usage'], 'r-s', label='Memory', linewidth=2)
ax5.plot(recent_metrics['timestamp'], recent_metrics['storage_usage'], 'g-^', label='Storage', linewidth=2)
ax5.set_xlabel('Time Step')
ax5.set_ylabel('Utilization (%)')
ax5.set_title('Resource Utilization Over Time')
ax5.legend()
ax5.grid(True, alpha=0.3)

# 6. Performance Summary
ax6 = axes[1, 2]
ax6.axis('off')
summary_text = f"""
PERFORMANCE SUMMARY

🎯 Performance Index: {tracker.get_performance_index():.1f}/100
🏥 Health Status: {tracker.get_health_status()}

📊 DATA PROCESSING
  • Load Time: {load_time:.3f}s
  • Prep Time: {prep_time:.3f}s

🤖 MODEL TRAINING
  • Training Time: {train_time:.3f}s
  • R² Score: {metrics['r2']:.4f}
  • RMSE: {metrics['rmse']:.4f}

⚡ ALLOCATION
  • Allocation Time: {alloc_time:.3f}s
  • Success Rate: {success_rate:.1f}%

💾 UTILIZATION
  • CPU: {util['cpu_utilization']:.1f}%
  • Memory: {util['memory_utilization']:.1f}%
  • Storage: {util['storage_utilization']:.1f}%
"""
ax6.text(0.05, 0.95, summary_text, transform=ax6.transAxes, fontsize=10,
        verticalalignment='top', fontfamily='monospace',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()

print("✅ Performance visualization generated!")
```
